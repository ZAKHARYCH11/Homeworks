>1.43. Наудачу взятый телефонный номер состоит из 5 цифр.
>Какова вероятность того, что в нем все цифры: а) различные;
>б) одинаковые; в) нечетные? Известно, что номер телефона не начинается
>с цифры ноль.

```python
nums = [i for i in range(10)]

p = [i for i in list(permutations(nums, 5)) if i[0] != 0]

print(f"а) {round(len(p) / (10**5 - 10**4), 4)}")

print(f"б) {round(9 / (10**5 - 10**4), 4)}")

print(f"в) {round(5**5 / (10**5 - 10**4), 4)}")
```

>1.48. В старинной игре в кости необходимо было для выигрыша
>получить при бросании трех игральных костей сумму очков, превосходящую
>10. Найти вероятности: а) выпадения l l очков; б) выигрыша.

```python
numbers = [i for i in range(1, 7)]

p = list(product(numbers, repeat=3))
count = 0
for i in p:
  if sum(i) == 11:
    count += 1
print(f"а) {round(count / 6**3, 4)}")

count = 0
for i in p:
  if sum(i) >= 11:
    count += 1
print(f"б) {round(count / 6**3, 4)}")
```
