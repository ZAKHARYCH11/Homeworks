>В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
from random import randint
import scipy.special as sp
```

```python
def getball(urn1, urn2):
  t = []
  for i in range(3):
    j = randint(0, len(urn1) - 1)
    t.append(urn1[j])
    del urn1[j]
  
  urn2 += t
  return urn2[randint(0, len(urn2) - 1)]
```

```python
urn1 = ['Б'] * 15 + ['Ч'] * 5
urn2 = ['Б'] * 6 + ['Ч']

N = 10**6
count = 0
for i in range(N):
  if getball(urn1.copy(), urn2.copy()) == 'Б':
    count += 1

print(round(count / N, 4))
```

```python
n = sp.comb(20, 3)
p0 = sp.comb(5, 3) / n
p1 = sp.comb(5, 2) * sp.comb(15, 1) / n
p2 = sp.comb(5, 1) * sp.comb(15, 2) / n
p3 = sp.comb(15, 3) / n

p = p0 * 6 / 10 + p1 * 7 / 10 + p2 * 8 /10 + p3 * 9 / 10

print(round(p, 4))
```
