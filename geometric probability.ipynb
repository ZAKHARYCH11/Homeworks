>Имеется синий квадрат со стороной 4. В центре квадрата расположен красный круг с радиусом 1. Стрелок при выстреле гарантировано попадает только в правую верхнюю четверть квадрата. Какова вероятность того, что он попадет в красный круг?

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle
import numpy as np
```

```python
def shots(a, r, n):
  x0, y0 = 2, 2
  ax = plt.subplots(figsize=(5, 5))
  ax = plt.gca()
  ax.add_patch(Rectangle((0, 0), a, a))
  ax.add_patch(Circle((x0, y0), r, facecolor='r'))
  
  x = np.random.uniform(0, a / 2, n)
  y = np.random.uniform(a / 2, a, n)
  plt.scatter(x, y, marker='x', c='black')
  
  inside = len([d for d in (x - x0) ** 2 + (y - y0) ** 2 if d <= r ** 2])
  prob = inside / n
  
  plt.show()
  print(f'В красный круг из {n} выстрелов попало {inside}. Доля попаданий {prob}.')
```

```python
a, r = 4, 1
shots(4, 1, 1000)
```

```python
p = (np.pi * r ** 2 / 4) / (a ** 2 / 4)
print(f'Теоретический расчет вероятности попадания {round(p, 4)}')
```
