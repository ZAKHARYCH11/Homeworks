```python
import matplotlib.pyplot as plt
import numpy as np
```

>Определить вероятность того, что сумма очков на двух брошенных игральных кубиках не менее 8.

```python
def dice_throw(n):
  a = np.random.randint(1, 7, n)
  b = np.random.randint(1, 7, n)
  count = np.sum((a + b) >= 8)
  
  return count / n
```

```python
classic_prob = 15 / 36

tests_numbers = [10**i for i in range(1, 7)]

probs = [dice_throw(i) for i in tests_numbers]

plt.figure(figsize=(10, 6))

plt.plot(tests_numbers, probs, 'r-*', label='Частотная вероятность')

plt.axhline(y=classic_prob, color='r', linestyle='--', label=f'Классическая вероятность {classic_prob:.4f}')

plt.xscale('log')
plt.xlabel('Число испытаний (n)')
plt.ylabel('Вероятность')
plt.legend()

plt.show()
```
