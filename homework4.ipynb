```python
print(f"1) {round(binom.pmf(5, 9, 0.8), 4)}")
print(f"2a) {round(binom.pmf(1, 9, 0.2) + binom.pmf(0, 9, 0.2), 4)}")
print(f"2б) {round(binom.pmf(1, 9, 0.2) + binom.pmf(0, 9, 0.2) + binom.pmf(2, 9, 0.2), 4)}")
print(f"2в) {round(1 - (binom.pmf(1, 9, 0.2) + binom.pmf(0, 9, 0.2)), 4)}")
print(f"2г) {round(binom.pmf(1, 9, 0.2) + binom.pmf(2, 9, 0.2), 4)}")
```

>Пример 2.11. По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.

```python
def local_moivre_laplace(m, n, p):
  q = 1 - p
  sigma = sqrt(n * p * q)
  x = (m - n * p) / sigma
  prob = norm.pdf(x) / sigma
  return prob

def integral_moivre_laplace(a, b, n, p):
  q = 1 - p
  sigma = sqrt(n * p * q)
  x1 = (a - n * p) / sigma
  x2 = (b - n * p) / sigma
  prob = norm.cdf(x2) - norm.cdf(x1)
  return prob
```

```python
print(f"а) {local_moivre_laplace(480, 1000, 0.5):.4f}")
print(f"б) {local_moivre_laplace(500, 1000, 0.5):.4f}")
print(f"в) {integral_moivre_laplace(480, 1000, 1000, 0.5):.4f}")
print(f"г) {integral_moivre_laplace(480, 520, 1000, 0.5):.4f}")
```
