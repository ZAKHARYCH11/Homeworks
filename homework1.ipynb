>Пример 1.8. В конкурсе по 5 номинациям участвуют 10 кинофильмов.
>Сколько существует вариантов распределения призов, если
>по каждой номинации установлены: а) различные призы; б) одинаковые
>призы?

```python
import itertools
```

```python
films = ["film_" + str(i) for i in range(1, 11)]
```

```python
p1_8a = list(itertools.product(films, repeat=5))
p1_8b = list(itertools.combinations_with_replacement(films, 5))

print(f"а) {len(p1_8a)}")
print(f"б) {len(p1_8b)}")
```

>Пример 1.9. Сколько существует семизначных чисел, состоящих
>из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры
>5 и 6 - по 2 раза?

```python
numbers = ["4", "5", "6"]
p1_9 = [s for s in itertools.product(numbers, repeat=7) if s.count("4") == 3 and s.count("5") == s.count("6") == 2]

print(len(p1_9))
```

>Пример 1.10. Буквы Т, Е, И, Я, Р, О написаны на отдельных
>карточках. Ребенок берет карточки в случайном порядке и прикладывает
>одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность
>того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

```python
letters = ["Т", "Е", "И", "Я", "Р", "О"]

print(f"а) {round(1 / len(list(itertools.permutations(letters, 3))), 4)}")

print(f"б) {round(1 / len(list(itertools.permutations(letters))), 4)}")
```

>Пример 1.11. Используя условие примера 1.10, найти вероятность
>того, что получится слово «АНАНАС», если на отдельных
>карточках написаны три буквы А, две буквы Н и одна буква С.

```python
letters = ["А", "Н", "А", "Н", "А", "С"]
letter_A = ["A", "A", "A"]
letter_N = ["Н", "Н"]
print(round(len(list(itertools.permutations(letter_A))) * len(list(itertools.permutations(letter_N))) / len(list(
    itertools.permutations(letters))), 4))
```

>Пример 1.16. За круглым столом рассаживаются 5 мужчин и
>5 женщин. Найти вероятность того, что: а) никакие два лица одного
>пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
>женщины образуют 5 супружеских пар.

```python
m = ["M"] * 5
w = ["W"] * 5
print(f"а) {round(2 * len(list(itertools.permutations(m)))**2 / len(list(itertools.permutations(m + w))), 4)}")
print(f"б) {round(len(list(itertools.permutations(m))) * 2**5 / len(list(itertools.permutations(m + w))), 4)}")
```

>Пример 1.17. В купейный вагон (9 купе по 4 места) семи пассажирам
>продано 7 билетов. Найти вероятности того, что пассажиры
>попали: а) в два купе; б) в семь купе; в) в три купе.

```python
c = ["C"] * 9
m = ["M"] * 36
n = len(list(itertools.combinations(m, 7)))
print(f"а) {round(len(list(itertools.combinations(c, 2))) * len(list(itertools.combinations(m[:8], 7))) / n, 4)}")

print(f"б) {round(len(list(itertools.combinations(c, 7))) * 4**7 / n, 4)}")

m3 = len(list(itertools.combinations(c, 3))) * (6 * len(list(itertools.combinations(m[:4], 4))) * len(list(itertools.combinations(m[:4], 2))) * len(list(itertools.combinations(m[:4], 1))) + 3 * len(list(itertools.combinations(m[:4], 3))) * len(list(itertools.combinations(m[:4], 3))) * len(list(itertools.combinations(m[:4], 1))) + 3 * len(list(itertools.combinations(m[:4], 3))) * len(list(itertools.combinations(m[:4], 2))) * len(list(itertools.combinations(m[:4], 2))))

print(f"в) {round(m3 / n, 4)}")
```
